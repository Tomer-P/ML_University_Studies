```python
#part a
import numpy as np
from tensorflow.keras.datasets import mnist
(X_train, y_train), (X_test, y_test) = mnist.load_data()
data = np.concatenate([X_train, X_test], axis=0).reshape(-1, 28 * 28)
labels = np.concatenate([y_train, y_test], axis=0)

import numpy.random
idx = numpy.random.RandomState(0).choice(70000, 11000)
train = data[idx[:10000], :].astype(int)
train_labels = labels[idx[:10000]]
test = data[idx[10000:], :].astype(int)
test_labels = labels[idx[10000:]]

k = 10
n = 1000
test_subset = test[:n]

train_limited = train[:1000]
train_labels_limited = train_labels[:1000]

train_sq = np.sum(train_limited**2, axis=1)
test_sq = np.sum(test_subset**2, axis=1, keepdims=True)
distances = test_sq + train_sq - 2 * np.dot(test_subset, train_limited.T)

neighbor_indices = np.argpartition(distances, k, axis=1)[:, :k]
neighbor_labels = train_labels_limited[neighbor_indices]

predictions = np.array([np.argmax(np.bincount(row)) for row in neighbor_labels])

accuracy = np.mean(predictions == test_labels[:n]) * 100
print(f"Final Accuracy: {accuracy:.2f}%")
```

```python
#part b
def knn_predict(train, train_labels, query_image, k):
  diff = train - query_image
  diff_squared = diff ** 2
  distances_squared = np.sum(diff_squared, axis=1)

  neighbor_indices = np.argsort(distances_squared)[:k]

  neighbor_labels = train_labels[neighbor_indices]

  prediction = np.bincount(neighbor_labels).argmax()

  return prediction
```

```python
#part c
import numpy as np
import matplotlib.pyplot as plt

k_values = range(1, 101)
n = 1000
n_train = 1000

train_subset = train[:n_train]
train_labels_subset = train_labels[:n_train]
test_subset = test[:n]

train_sq = np.sum(train_subset**2, axis=1)
test_sq = np.sum(test_subset**2, axis=1, keepdims=True)
distances = test_sq + train_sq - 2 * np.dot(test_subset, train_subset.T)

neighbor_indices = np.argsort(distances, axis=1)[:, :100]
neighbor_labels = train_labels_subset[neighbor_indices]

accuracies = []
for k in k_values:
    current_k_labels = neighbor_labels[:, :k]

    predictions = np.array([np.argmax(np.bincount(row)) for row in current_k_labels])

    acc = np.mean(predictions == test_labels[:n]) * 100
    accuracies.append(acc)

plt.figure(figsize=(12, 6))
plt.plot(k_values, accuracies, color='#E14D8C', linewidth=2)
plt.title(f'Accuracy as a function of k (n_train={n_train})', fontsize=14)
plt.xlabel('k', fontsize=12)
plt.ylabel('Accuracy (%)', fontsize=12)
plt.grid(True, linestyle='--', alpha=0.7)

best_idx = np.argmax(accuracies)
best_k = list(k_values)[best_idx]
print(f"\nDone! Best k is {best_k} with {accuracies[best_idx]:.2f}% accuracy.")

plt.show()
```

```python
#part d
import numpy as np
import matplotlib.pyplot as plt

n_values = range(100, 5001, 100)
accuracies_n = []
num_test = 1000

test_subset = test[:num_test]
train_max = train[:5000]
train_labels_max = train_labels[:5000]

test_sq = np.sum(test_subset**2, axis=1, keepdims=True)
train_sq = np.sum(train_max**2, axis=1)
inner_prod = np.dot(test_subset, train_max.T)

all_distances = test_sq + train_sq - 2 * inner_prod

for n in n_values:
    current_dists = all_distances[:, :n]

    indices = np.argmin(current_dists, axis=1)
    preds = train_labels_max[indices]

    acc = np.mean(preds == test_labels[:num_test]) * 100
    accuracies_n.append(acc)

best_idx = np.argmax(accuracies_n)
print(f"Done! Best n: {list(n_values)[best_idx]}, Best Accuracy: {accuracies_n[best_idx]:.2f}%")

plt.figure(figsize=(10, 6))

plt.plot(n_values, accuracies_n, color='#E14D8C', linestyle='-', linewidth=2, marker='o', markersize=3, alpha=0.8)

plt.title('Accuracy as a function of n (k=1)', fontsize=14, fontweight='bold')
plt.xlabel('Training Set Size (n)', fontsize=12)
plt.ylabel('Accuracy (%)', fontsize=12)

plt.grid(True, which='both', linestyle='--', alpha=0.5)

plt.ylim(min(accuracies_n) - 2, 100)

plt.show()
```
